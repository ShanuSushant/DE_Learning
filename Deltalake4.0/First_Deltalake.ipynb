```python
from delta.tables import DeltaTable, IdentityGenerator
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

## **Create Delta Table**
#### **Using SQL API**

```python
%sql
create table deltalakesushant.default.first_table
(
    id INT,
    salary INT
)
```

#### **Using Delta API**

```python
from delta.tables import DeltaTable, IdentityGenerator
```

```python
DeltaTable.createIfNotExists(spark)\
    .tableName("deltalakesushant.default.first_table_api")\
    .addColumn("id","INT")\
    .addColumn("salary","INT")\
    .execute()
```

```python
%sql
DROP TABLE deltalakesushant.default.first_table_api
```

### **Generated Columns**
#### **Identity Columns:**

```python
 DeltaTable.createIfNotExists(spark)\
    .tableName("deltalakesushant.default.first_table_api")\
    .addColumn("id", dataType=LongType(), generatedAlwaysAs=IdentityGenerator())\
    .addColumn("salary",dataType=IntegerType())\
    .addColumn("name", dataType=StringType())\
    .execute()
```

```python
%sql
INSERT INTO deltalakesushant.default.first_table_api(salary,name)
VALUES
(10000,'Shyam'),
(20000,'Ravi'),
(30000,'Rajesh'),
(40000,'Rakesh')
```

```python
%sql
select * from deltalakesushant.default.first_table_api
```

#### **Computed Columns:**

```python
%sql
DROP TABLE deltalakesushant.default.first_table_api2
```

```python
DeltaTable.createIfNotExists(spark)\
    .tableName('deltalakesushant.default.first_table_api2')\
    .addColumn('id',dataType=LongType(),generatedAlwaysAs=IdentityGenerator())\
    .addColumn('name',dataType=StringType())\
    .addColumn('salary',dataType=IntegerType())\
    .addColumn('salaryTax',dataType=LongType(),generatedAlwaysAs= "CAST((salary*0.7) AS BIGINT)")\
    .execute()
```

```python
%sql
INSERT INTO deltalakesushant.default.first_table_api2(name,salary)
VALUES
('Ram',1000),
('Shyam',2000),
('Joe',100),
('Rai',5000)
```

```python
%sql
SELECT * FROM deltalakesushant.default.first_table_api2
```
