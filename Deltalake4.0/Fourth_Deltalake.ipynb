### Upsert Operation:
###### _If Exists -> Update, If Not Exists -> Insert_

```python
from delta.tables import DeltaTable
```

```python
data = [(1,80000,'xx','Y'),(10,200,'bb','Y'),(9,300,'cc','N')]
df = spark.createDataFrame(data,['id','salary','name','isactive'])
display(df)
```

```python
## create delta object
delta_obj = DeltaTable.forPath(spark,"/Volumes/deltalakesushant/default/deltalakevol/dmlSink/")

## create merge statement using upsert logic
delta_obj.alias("target").merge(df.alias("source"),"target.id = source.id")\
    .whenMatchedUpdateAll()\
    .whenNotMatchedInsertAll()\
    .execute()
```
