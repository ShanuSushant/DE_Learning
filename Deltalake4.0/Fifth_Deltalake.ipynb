### Schema Level Changes:

```python
data = [(1,80000,'xx','Y'),(10,200,'bb','Y'),(9,300,'cc','N')]
df = spark.createDataFrame(data,['id','salary','name','isactive'])
display(df)
```

```python
df.write.format("delta")\
    .mode("append")\
    .save("/Volumes/deltalakesushant/default/deltalakevol/schemaSink/")
```

```python
%sql
-- Enabling  column mapping on the delta table
ALTER TABLE DELTA.`/Volumes/deltalakesushant/default/deltalakevol/schemaSink/`
SET TBLPROPERTIES 
(   
    'delta.minReaderVersion' = '2',
    'delta.minWriterVersion' = '5',
    'delta.columnMapping.mode' = 'name'
);

-- Rename the column name
ALTER TABLE DELTA.`/Volumes/deltalakesushant/default/deltalakevol/schemaSink/`
RENAME COLUMN name to customer_name;
```
