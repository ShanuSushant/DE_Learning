#### CLONE DATA:
> ###### There are two types of cloning of data-
> - Shallow Cloning - Clones the data without metadata
> - Deep Cloning - Clones data along with the metadata (by default cloning)


```python
%sql
CREATE TABLE deltalakesushant.default.cloned_table
CLONE DELTA.`/Volumes/deltalakesushant/default/deltalakevol/dmlSink/` VERSION AS OF 3
```

```python
%sql
SELECT * FROM deltalakesushant.default.cloned_table
```

```python
%sql
CREATE TABLE deltalakesushant.default.shallow_cloned_table
SHALLOW CLONE deltalakesushant.default.cloned_table
```

```python
%sql
SELECT * FROM deltalakesushant.default.shallow_cloned_table
```
